```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
```

```python
# 1. Load the breast cancer dataset
data = load_breast_cancer()
X, y = data.data, data.target 
```

```python
def build_model(input_dim):
    model = Sequential()
    model.add(Dense(units=10, input_dim=input_dim, activation='sigmoid'))
    model.add(Dense(units=5, activation='sigmoid'))
    model.add(Dense(units=1, activation='sigmoid'))

    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
# Train on raw (unscaled) data
print("Training on Unscaled Data")
model_unscaled = build_model(X.shape[1])
model_unscaled.fit(X, y, epochs=40, batch_size=32)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
print("\n Training on Scaled Data")
model_scaled = build_model(X_scaled.shape[1])
model_scaled.fit(X_scaled, y, epochs=40, batch_size=32)
```
